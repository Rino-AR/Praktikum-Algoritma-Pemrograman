```python
class Stack: # pop = menghapus elemen
  def __init__(self):
    self.stack = []

  def push(self, item):
    self.stack.append(item)

  def pop(self):
    if len (self.stack)== 0:
      return "stack kosong"
    return self.stack.pop()

#contoh
s = Stack()
s.push(5)
s.push(10)
s.push(15)
print(s.pop())
print (s.stack)
```

```python
class Stack:
  def __init__(self):
    self.stack =[]

  def push(self, item):
    self.stack.append(item)

  def peek(self): #peek = ngintip data terakhir / teratas
    if len (self.stack) == 0:
      return "stack kosong"
    return self.stack[-1]

# Contoh
s = Stack()
s.push(5)
s.push(10)
print(s.peek()) # Output: 10
print(s.stack) #output tetap [5,10]
```

```python
class Stack:
  def __init__(self):
    self.stack = []

  def is_empty(self):           # is_empty -> mengecek apakah satck kosong
    return len(self.stack) == 0

  def push(self, item):
    self.stack.append(item)

# Contoh penggunaan
s = Stack()
print(s.is_empty()) # Output: True
s.push (7)
print(s.is_empty()) # Output: False
```

```python
#  Implementasi Stack dengan List Python - Fitur "Undo" dalam Teks Editor (pop())
class TextEditor:
  def __init__(self):
    self.history = []

  def write(self, text):
    self.history.append(text)

  def undo (self):
    if self.history:
      return f"Undo: Menghapus '{self.history.pop()}'"
    return "Tidak ada yang bisa di-undo"

# Contoh
editor = TextEditor ()
editor.write("Halo, ini teks pertama.")
editor.write("Ini teks kedua.")
editor.write("Ini teks ketiga.")
print (editor.undo())
print (editor.undo ())
```

**Latihan 1**:
Buatlah program "Undo Perintah Kalkulator", di mana setiap operasi matematika
yang dilakukan disimpan dalam Stack. Jika pengguna menekan Undo, operasi
terakhir dibatalkan. Gunakan dua digit terakhir NPM sebagai angka awal. Lakukan
minimal 3 operasi matematika sebelum menguji fitur Undo.

```python
class Kalkulator:
    def __init__(self, hasil_akhir):
        self.value = hasil_akhir
        self.history = []

    def tambah(self, angka):
        self.history.append(self.value)
        self.value += angka

    def kurang(self, angka):
        self.history.append(self.value)
        self.value -= angka

    def kali(self, angka):
        self.history.append(self.value)
        self.value *= angka

    def bagi(self, angka):
        if angka != 0:
            self.history.append(self.value)
            self.value /= angka
        else:
            print("Tidak bisa membagi dengan nol!")

    def undo(self):
        if self.history:
            self.value = self.history.pop()
        else:
            print("Tidak ada operasi yang bisa di undo")

    def hasil(self):
        return self.value

angka_awal = int(input("Masukkan 2 digit terakhir NPM sebagai angka awal: "))
k = Kalkulator(angka_awal)

while True:
    print('''\nKalkulator
1. Tambah
2. Kurang
3. Kali
4. Bagi
5. Undo
6. Keluar''')

    try:
        pilihan = int(input("Masukkan pilihan anda: "))
        if pilihan == 1:
            angka = int(input("Masukkan angka yang ingin ditambah: "))
            k.tambah(angka)
            print("Hasil:", k.hasil())
        elif pilihan == 2:
            angka = int(input("Masukkan angka yang ingin dikurang: "))
            k.kurang(angka)
            print("Hasil:", k.hasil())
        elif pilihan == 3:
            angka = int(input("Masukkan angka yang ingin dikali: "))
            k.kali(angka)
            print("Hasil:", k.hasil())
        elif pilihan == 4:
            angka = int(input("Masukkan angka yang ingin dibagi: "))
            k.bagi(angka)
            print("Hasil:", k.hasil())
        elif pilihan == 5:
            k.undo()
            print("Hasil setelah undo:", k.hasil())
        elif pilihan == 6:
            print("Terima kasih telah menggunakan kalkulator!")
            break
        else:
            print("Masukkan angka 1-6!")
    except ValueError:
        print("Masukkan angka yang benar.")
```

```python
#Implementasi Stack dengan collections.deque - Navigasi History Browser (push())
from collections import deque

class BrowserHistory:
  def __init__(self):
    self.history = deque()

  def visit(self, url):
    self.history.append(url)
    print(f"Mengunjungi: {url}")

  def back(self):
    if self.history:
      print(f"Kembali ke: {self.history.pop()}")
    else:
      print("Tidak ada riwayat!")

# Contoh Penggunaan
browser = BrowserHistory()
browser.visit("google.com")
browser.visit("youtube.com")
browser.visit("github.com")

browser.back ()
browser.back ()
```

**Latihan 2**:
Buatlah program "Navigasi History Aplikasi Mobile", di mana fitur aplikasi yang
dikunjungi disimpan dalam Stack.- Gunakan huruf pertama bulan lahir + dua digit
terakhir NPM sebagai nama fitur aplikasi. Simpan minimal 3 fitur sebelum menguji
tombol Back.

```python
from collections import deque

class AppHistory:
    def __init__(self):
        self.history = deque()

    def visit(self, fitur):
        self.history.append(fitur)
        print(f"Mengunjungi: {fitur}")

    def back(self):
        if self.history:
            print(f"Kembali ke: {self.history.pop()}")
        else:
            print("Tidak ada riwayat!")

app = AppHistory()
app.visit("A61")
app.visit("A62")
app.visit("A63")

app.back()
app.back()
app.back()
app.back()
```

```python
# Implementasi Stack dengan queue.LifoQueue - Sistem Pencadangan File (peek())
from queue import LifoQueue

class BackupSystem:
  def __init__(self):
    self.backups = LifoQueue ()

  def backup (self, version):
    self.backups.put(version)

  def latest_backup (self):
    if not self.backups.empty ():
      return f"Versi terbaru: {self.backups.queue[-1]}"
    return "Tidak ada cadangan!"

# Contoh Penggunaan
backup_system = BackupSystem()
backup_system.backup("Backup_1")
backup_system.backup ("Backup_2")
backup_system.backup("Backup_3")
print (backup_system.latest_backup())
```

**Latihan 3**=
Buatlah program "Sistem Penyimpanan Draft Email", di mana email yang belum
dikirim disimpan dalam Stack. Gunakan dua huruf pertama nama mahasiswa + dua
digit terakhir NPM sebagai subjek email. Simpan minimal 3 draft email sebeluт
menguji fitur "Lihat Draft Terakhir".

```python
#LATIHAN 3
from queue import LifoQueue

class EmailDraftSystem:
    def __init__(self):
        self.drafts = LifoQueue()

    def simpan_draft(self, draft):
        self.drafts.put(draft)

    def lihat_draft_terakhir(self):
        if not self.drafts.empty():
            return f"Draft Terakhir: {self.drafts.queue[-1]}"
        return "Tidak ada draft!"

# Contoh penggunaan
sistem_email = EmailDraftSystem()
sistem_email.simpan_draft("NA59")
sistem_email.simpan_draft("NA60")
sistem_email.simpan_draft("RI61")
print(sistem_email.lihat_draft_terakhir())
```

```python
# Implementasi Stack dengan ООР - Manajemen Antrian Pelanggan (isEmpty())
class CustomerService:
  def __init__ (self):
    self.tickets = []

  def add_ticket(self, ticket):
    self.tickets.append (ticket)

  def process_ticket(self):
    if not self.is_empty():
      return f"Memproses tiket: {self.tickets.pop() }"
    return "Tidak ada tiket yang harus diproses."

  def is_empty(self):
    return len(self.tickets) == 0

# Contoh
service = CustomerService()
service.add_ticket("Tiket 001")
service.add_ticket("Tiket 002")
print (service.process_ticket())
print (service.is_empty ())
```

**Latihan 4**=
Buatlah program "Sistem Antrean Tiket Bioskop", di mana pelanggan membeli tiket
film dan antrean disimpan dalam Stack. Gunakan huruf pertama bulan lahir + dua
digit terakhir NPM sebagai nama film. Beli minimal 3 tiket sebelum menguji sistem
pemrosesan tiket.

```python
#LATIHAN 4
class CinemaQueue:
    def __init__(self):
        self.tickets = []

    def beli_tiket(self, tiket):
        self.tickets.append(tiket)

    def proses_tiket(self):
        if not self.is_empty():
            return f"Memproses tiket: {self.tickets.pop()}"
        return "Tidak ada tiket yang harus diproses."

    def is_empty(self):
        return len(self.tickets) == 0

# Contoh penggunaan
bioskop = CinemaQueue()
bioskop.beli_tiket("C3")
bioskop.beli_tiket("B2")
bioskop.beli_tiket("A1")
print(bioskop.proses_tiket())
```

```python
class Gudang:
    def __init__(self, kapasitas):
        self.kapasitas = kapasitas
        self.barang = []

    def tambah_barang(self, item):
        if len(self.barang) < self.kapasitas:
            self.barang.append(item)
            print(f"Barang '{item}' ditambahkan ke gudang.")
        else:
            print("Gudang penuh!")

    def ambil_barang(self):
        if self.barang:
            return f"Mengambil barang: {self.barang.pop()}"
        return "Gudang kosong!"

kapasitas_gudang = int(61) #NPM last number = 61

gudang = Gudang(kapasitas_gudang)
gudang.tambah_barang("Meja")
gudang.tambah_barang("Kursi")
gudang.tambah_barang("Kayu")
print(gudang.ambil_barang())
```
